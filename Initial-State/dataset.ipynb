```python
import os
import json
import pandas as pd

# Path to the dataset folder
base_path = "data/data/Dataset-LinkSocial/3.profile.data"

# Collect all profile data
all_profiles = []

for user_folder in os.listdir(base_path):
    user_path = os.path.join(base_path, user_folder)
    if os.path.isdir(user_path):
        for file in os.listdir(user_path):
            if file.endswith('.json'):
                with open(os.path.join(user_path, file), 'r', encoding='utf-8') as f:
                    data = json.load(f)
                    # Add platform info based on filename
                    if 'twitter' in file.lower():
                        data['platform'] = 'twitter'
                    elif 'instagram' in file.lower():
                        data['platform'] = 'instagram'
                    elif 'google' in file.lower():
                        data['platform'] = 'googleplus'
                    all_profiles.append(data)

# Convert to DataFrame
df = pd.DataFrame(all_profiles)
print(df.head())
print(f"Total profiles: {len(df)}")
```

```python
import pandas as pd

# ============================================================
# โหลดแต่ละ Table
# ============================================================

# วิธี 1: จาก Parquet (เร็วกว่า)
images = pd.read_parquet('data/normalized/images.parquet')
metadata = pd.read_parquet('data/normalized/image_metadata.parquet')
quality = pd.read_parquet('data/normalized/image_quality.parquet')
faces = pd.read_parquet('data/normalized/image_faces.parquet')
labels = pd.read_parquet('data/normalized/image_labels.parquet')
captions = pd.read_parquet('data/normalized/image_captions.parquet')
embeddings = pd.read_parquet('data/normalized/image_embeddings.parquet')

# วิธี 2: จาก CSV
images = pd.read_csv('data/normalized/images.csv')

# วิธี 3: จาก SQLite
import sqlite3
conn = sqlite3.connect('data/normalized/image_database.sqlite')
images = pd.read_sql('SELECT * FROM images', conn)

# ============================================================
# ดูข้อมูล
# ============================================================

# ดู 5 แถวแรก
print(images.head())

# ดูทุก Table พร้อมกัน
for name in ['images', 'image_metadata', 'image_quality', 'image_faces', 'image_labels', 'image_captions']:
    df = pd.read_parquet(f'data/normalized/{name}.parquet')
    print(f"\n=== {name} ({len(df)} rows) ===")
    print(df.head(3))

# ============================================================
# JOIN ทุก Table เข้าด้วยกัน
# ============================================================

merged = images.merge(metadata, on='item_id') \
               .merge(quality, on='item_id') \
               .merge(faces, on='item_id') \
               .merge(labels, on='item_id') \
               .merge(captions, on='item_id', how='left')

print(merged.head())

# ============================================================
# Filter ข้อมูลคุณภาพดี
# ============================================================

quality_images = merged[
    (merged['is_blurry'] == False) & 
    (merged['is_duplicate'] == False)
]
print(f"Quality images: {len(quality_images)}")
```
