```python
# ดู features ทั้ง 17
feature_cols = ['username_jaro','username_token_sort','username_levenshtein','username_both_empty',
                'fullname_jaro','fullname_token_sort','fullname_levenshtein','fullname_both_empty',
                'bio_jaro','bio_token_sort','bio_levenshtein','bio_both_empty',
                'bio_tfidf_cosine','url_exact_match','url_domain_match','same_platform','location_sim']
print('=== feature distribution (SAME_PERSON) ===')
df[df['label']==1][feature_cols].describe().round(3)
```

```python
df['platform_pair'] = df.apply(
    lambda r: '_'.join(sorted([str(r['platform_a']), str(r['platform_b'])])), axis=1
)
print('=== platform_pair vs label ===')
df.groupby('platform_pair')['label'].value_counts().unstack(fill_value=0)
```
