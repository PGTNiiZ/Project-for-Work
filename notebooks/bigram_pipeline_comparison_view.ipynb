# Baseline vs Location Comparison
เปรียบเทียบผล `bigram pipeline` เดิม กับเวอร์ชันที่เพิ่ม location features


```python
import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 220)
```

```python
path = 'data/processed/bigram_pipeline_results_with_location/bigram_vs_location_comparison.csv'
df = pd.read_csv(path)
print('rows:', len(df))
df
```

```python
# ตารางดู baseline/with-location แบบอ่านง่าย
tbl = df[['pair_name','model_norm','top1_accuracy_baseline','top1_accuracy_with_location','delta_top1_accuracy','candidate_recall_baseline','candidate_recall_with_location','delta_candidate_recall']]
tbl.sort_values(['pair_name','model_norm'])
```

```python
# Pivot สำหรับดู delta ตามคู่แพลตฟอร์ม
p_delta_acc = df.pivot(index='pair_name', columns='model_norm', values='delta_top1_accuracy').sort_index()
p_delta_rec = df.pivot(index='pair_name', columns='model_norm', values='delta_candidate_recall').sort_index()
print('Delta Top1 Accuracy')
display(p_delta_acc)
print('Delta Candidate Recall')
display(p_delta_rec)
```

```python
# กราฟ baseline vs with-location ของ Top1 Accuracy
fig, ax = plt.subplots(figsize=(10,4))
x = range(len(df))
labels = (df['pair_name'] + ' | ' + df['model_norm']).tolist()
ax.plot(x, df['top1_accuracy_baseline'], marker='o', label='baseline')
ax.plot(x, df['top1_accuracy_with_location'], marker='o', label='with_location')
ax.set_xticks(list(x))
ax.set_xticklabels(labels, rotation=25, ha='right')
ax.set_ylabel('top1_accuracy')
ax.set_title('Top1 Accuracy: Baseline vs With Location')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# กราฟ delta
fig, axes = plt.subplots(1, 2, figsize=(14,4))
axes[0].bar(labels, df['delta_top1_accuracy'])
axes[0].set_title('Delta Top1 Accuracy (with - baseline)')
axes[0].tick_params(axis='x', rotation=25)

axes[1].bar(labels, df['delta_candidate_recall'])
axes[1].set_title('Delta Candidate Recall (with - baseline)')
axes[1].tick_params(axis='x', rotation=25)

plt.tight_layout()
plt.show()
```
