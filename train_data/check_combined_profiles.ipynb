# Explore: combined_profiles.csv & cleaned_social_data.csv

```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE = Path('/Users/tm/Documents/GitHub/Project-for-Work/data_for_project/')

df_combined = pd.read_csv(BASE / 'combined_profiles.csv')
df_social   = pd.read_csv(BASE / 'cleaned_social_data.csv')

print('combined_profiles :', df_combined.shape)
print('cleaned_social_data:', df_social.shape)
```

```python
df_combined.head(10)
```

## 3. Null / Missing values

```python
null_stats = pd.DataFrame({
    'null_count': df_combined.isnull().sum(),
    'null_%': (df_combined.isnull().mean() * 100).round(1)
})
null_stats
```

## 4. ค่าสถิติเบื้องต้น (ความยาว string)

```python
str_cols = ['userName', 'fullName', 'bio', 'location', 'externalUrl']
len_stats = pd.DataFrame({
    col: df_combined[col].dropna().str.len().describe()
    for col in str_cols
})
len_stats.round(1)
```

## 5. user_folder distribution

```python
folder_counts = df_combined['user_folder'].value_counts()
print(f'Unique user_folder: {folder_counts.nunique():,}')
print(f'Folders with >= 2 profiles: {(folder_counts >= 2).sum():,}')
print(f'Max profiles per folder:    {folder_counts.max()}')
print()
print('Distribution of profiles per folder:')
folder_counts.value_counts().sort_index().head(10)
```

## 6. Folders ที่มี profiles มากที่สุด (Top 20)

```python
top_folders = folder_counts.head(20).reset_index()
top_folders.columns = ['user_folder', 'profile_count']
top_folders
```

## 7. ดู profiles ใน folder เดียวกัน (same person)

```python
# เปลี่ยน folder_name เพื่อดู group อื่น
folder_name = folder_counts.index[0]
print(f'Showing folder: {folder_name}')
df_combined[df_combined['user_folder'] == folder_name][['userName','fullName','bio','location','user_folder']]
```

## 8. เปรียบเทียบ cleaned_social_data vs combined_profiles

```python
print('=== cleaned_social_data ===')
print(df_social.columns.tolist())
display(df_social.isnull().sum().rename('null_count').to_frame())

print('\n=== combined_profiles ===')
print(df_combined.columns.tolist())
display(df_combined.isnull().sum().rename('null_count').to_frame())
```
